```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
plt.style.use('seaborn-v0_8-darkgrid')
```

```python
with open('data/logs-insights-results.csv', 'r') as file: # Logs from processing 50 files 3 times, usually the required BAM is the same but not always
    lines = file.readlines()

extracted_values = []
for line in lines:
    if "Max memory needed for sorting = " in line:
        value = line.split('=')[1].strip()
        extracted_values.append(int(value))

print(f"Avg memory needed for sorting of 50 files: {np.mean(extracted_values) / 1024**3:.2f} GiB")
print(f"Max memory needed for sorting of 50 files: {np.max(extracted_values) / 1024**3:.2f} GiB")
```

```python
print(f"Avg memory needed for sorting of 50 files (95 percentile): {np.percentile(extracted_values, 95) / 1024**3:.2f} GiB")
```

```python
pd.Series(extracted_values).plot.box()
```
